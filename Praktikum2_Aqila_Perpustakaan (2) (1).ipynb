## Bagian B — Seleksi dan filter (Latihan 2, berpasangan)
**Tujuan:** mengambil bagian data yang relevan dengan pertanyaan analisis.

````python
# DEMO B1 — memilih kolom

# Satu kolom -> Series
df['durasi_menit'].head()

# Beberapa kolom -> DataFrame
df[['id_kunjungan', 'prodi', 'durasi_menit']].head()
````

````python
# DEMO B2 — loc (berbasis label) vs iloc (berbasis posisi)

# loc: mengambil berdasarkan nama kolom
print(df.loc[0, 'prodi'])

# iloc: mengambil berdasarkan posisi kolom
# kolom ke-4 adalah angkatan
print(df.iloc[0, 4])

# loc: mengambil baris 0 sampai 4 dan beberapa kolom
df.loc[0:4, ['id_kunjungan', 'prodi', 'durasi_menit']]
````

````python
# DEMO B3 — filter dengan kondisi (boolean mask)

mask = df['durasi_menit'] > 60

print('Kunjungan dengan durasi di atas 60 menit:', mask.sum())

df[mask].head()
````

````python
# DEMO B4 — menggabungkan kondisi

# Kunjungan dengan durasi > 60 menit DAN jumlah pinjaman >= 2
hi = df[
    (df['durasi_menit'] > 60) &
    (df['jumlah_pinjam'] >= 2)
]

print('Durasi > 60 menit dan jumlah pinjam >= 2:', hi.shape)

# Kunjungan yang menggunakan Ruang Baca atau Ruang Diskusi
ruang = df[
    df['jenis_ruang'].isin(['Ruang Baca', 'Ruang Diskusi'])
]

print('Kunjungan di Ruang Baca atau Ruang Diskusi:', ruang.shape)
````

````python
# TODO B5 — Data kunjungan dengan kepuasan rendah

# a. Ambil semua kunjungan dengan kepuasan < 3
kepuasan_rendah = df[df['kepuasan'] < 3]

print("Kunjungan dengan kepuasan < 3:")
print(kepuasan_rendah)

# b. Jumlah kunjungan dan rata-rata durasi
print("\nJumlah kunjungan:", len(kepuasan_rendah))
print("Rata-rata durasi:", kepuasan_rendah['durasi_menit'].mean())

# c. Ambil kunjungan dengan jumlah pinjam >= 2
pinjam_banyak = df[df['jumlah_pinjam'] >= 2]

print("\nKunjungan dengan jumlah pinjam >= 2:")
print(pinjam_banyak)
````

````python
# TODO B6 — pertanyaan bisnis

# Tampilkan 10 kunjungan dengan durasi paling lama
# dengan jumlah pinjaman minimal 1

hasil_b6 = df[df['jumlah_pinjam'] >= 1]

hasil_b6 = hasil_b6.sort_values(
    'durasi_menit',
    ascending=False
)

print("10 kunjungan dengan durasi paling lama:")
print(
    hasil_b6[
        ['id_kunjungan', 'prodi', 'durasi_menit',
         'jumlah_pinjam', 'jenis_ruang', 'kepuasan']
    ].head(10)
)
````

## Bagian C — Kolom turunan dan pengurutan

````python
# DEMO C1 — membuat kolom baru

# Membuat kolom kategori durasi
df['kategori_durasi'] = np.where(
    df['durasi_menit'] > 60,
    'Lama',
    'Singkat'
)

df[['durasi_menit', 'kategori_durasi']].head()
````

````python
# DEMO C2 — kolom tanggal

df['tanggal'] = pd.to_datetime(df['tanggal'])

df['bulan'] = df['tanggal'].dt.month

df['nama_hari'] = df['tanggal'].dt.day_name()

df[['tanggal', 'bulan', 'nama_hari']].head()
````

````python
# TODO C3

# a. Buat kolom kelas_durasi
# Lama jika durasi > 60 menit, selain itu Singkat
df['kelas_durasi'] = np.where(
    df['durasi_menit'] > 60,
    'Lama',
    'Singkat'
)

# b. Hitung jumlah tiap kelas
print("Jumlah berdasarkan kelas durasi:")
print(df['kelas_durasi'].value_counts())

# c. Tampilkan 5 kunjungan dengan durasi terbesar
print("\n5 kunjungan dengan durasi terbesar:")

print(
    df.sort_values(
        'durasi_menit',
        ascending=False
    )[
        ['id_kunjungan', 'prodi', 'durasi_menit',
         'jumlah_pinjam', 'jenis_ruang']
    ].head(5)
)
````

## Bagian D — Agregasi dengan groupby
Pola yang akan Anda pakai sepanjang semester: **split → apply → combine**.

````python
# DEMO D1 — agregasi sederhana

df.groupby('prodi')['durasi_menit'].mean().round(2)
````

````python
# DEMO D2 — beberapa agregasi sekaligus

ringkas = (
    df.groupby('prodi')
      .agg(
          jumlah_kunjungan=('id_kunjungan', 'count'),
          rata_rata_durasi=('durasi_menit', 'mean'),
          rata_rata_pinjam=('jumlah_pinjam', 'mean'),
          rata_rata_kepuasan=('kepuasan', 'mean')
      )
      .round(2)
      .sort_values('jumlah_kunjungan', ascending=False)
)

ringkas
````

````python
# DEMO D3 — dua kunci pengelompokan

df.groupby(
    ['prodi', 'jenis_ruang']
)['durasi_menit'].mean().unstack(fill_value=0)
````

````python
# c. Jumlah kunjungan per bulan

# Pastikan kolom tanggal menjadi datetime
df['tanggal'] = pd.to_datetime(df['tanggal'])

kunjungan_bulan = (
    df.groupby(df['tanggal'].dt.month)
      .size()
      .sort_values(ascending=False)
)

print("\nJumlah kunjungan per bulan:")
print(kunjungan_bulan)
````
