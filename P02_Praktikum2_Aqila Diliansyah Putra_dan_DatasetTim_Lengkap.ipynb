**Tujuan:** mengambil bagian data yang relevan dengan pertanyaan analisis.

````python
# DEMO B1 — memilih kolom
# Satu kolom menghasilkan Series.
df['durasi_menit'].head()

# Beberapa kolom menghasilkan DataFrame.
df[['id_kunjungan', 'prodi', 'durasi_menit']].head()
````

````python
# TODO B5 — kunjungan dengan kepuasan rendah
# a. Ambil semua kunjungan dengan kepuasan < 3.
kepuasan_rendah = df[df['kepuasan'] < 3]
print('Kunjungan dengan kepuasan < 3:')
display(kepuasan_rendah)

# b. Hitung jumlah kunjungan dan rata-rata durasinya.
print('\nJumlah kunjungan:', len(kepuasan_rendah))
print('Rata-rata durasi:', kepuasan_rendah['durasi_menit'].mean())

# c. Ambil kunjungan dengan jumlah pinjam >= 2.
pinjam_banyak = df[df['jumlah_pinjam'] >= 2]
print('\nKunjungan dengan jumlah pinjam >= 2:')
display(pinjam_banyak)
````

````python
# TODO B6 — pertanyaan bisnis
# Tampilkan 10 kunjungan berdurasi paling lama dengan jumlah pinjaman minimal 1.
hasil_b6 = (
    df[df['jumlah_pinjam'] >= 1]
      .sort_values('durasi_menit', ascending=False)
)

print('10 kunjungan dengan durasi paling lama:')
display(
    hasil_b6[
        ['id_kunjungan', 'prodi', 'durasi_menit',
         'jumlah_pinjam', 'jenis_ruang', 'kepuasan']
    ].head(10)
)
````

## Bagian C — Kolom turunan dan pengurutan

````python
# DEMO C1 — membuat kolom kategori durasi
df['kategori_durasi'] = np.where(
    df['durasi_menit'] > 60,
    'Lama',
    'Singkat'
)

df[['durasi_menit', 'kategori_durasi']].head()
````

````python
# DEMO C2 — kolom tanggal
df['tanggal'] = pd.to_datetime(df['tanggal'])
df['bulan'] = df['tanggal'].dt.month
df['nama_hari'] = df['tanggal'].dt.day_name()

df[['tanggal', 'bulan', 'nama_hari']].head()
````

````python
# TODO C3
# a. Kategorikan durasi: Lama jika >60 menit, selain itu Singkat.
df['kelas_durasi'] = np.where(
    df['durasi_menit'] > 60,
    'Lama',
    'Singkat'
)

# b. Hitung jumlah tiap kelas durasi.
print('Jumlah berdasarkan kelas durasi:')
print(df['kelas_durasi'].value_counts())

# c. Tampilkan 5 kunjungan dengan durasi terbesar.
print('\n5 kunjungan dengan durasi terbesar:')
display(
    df.sort_values('durasi_menit', ascending=False)[
        ['id_kunjungan', 'prodi', 'durasi_menit',
         'jumlah_pinjam', 'jenis_ruang']
    ].head(5)
)
````

## Bagian D — Agregasi dengan `groupby`
Pola analisis yang digunakan: **split → apply → combine**.

````python
# DEMO D1 — rata-rata durasi per program studi
df.groupby('prodi')['durasi_menit'].mean().round(2)
````

````python
# DEMO D2 — beberapa agregasi sekaligus
ringkas = (
    df.groupby('prodi')
      .agg(
          jumlah_kunjungan=('id_kunjungan', 'count'),
          rata_rata_durasi=('durasi_menit', 'mean'),
          rata_rata_pinjam=('jumlah_pinjam', 'mean'),
          rata_rata_kepuasan=('kepuasan', 'mean')
      )
      .round(2)
      .sort_values('jumlah_kunjungan', ascending=False)
)

ringkas
````

````python
# DEMO D3 — pengelompokan berdasarkan prodi dan jenis ruang
df.groupby(
    ['prodi', 'jenis_ruang']
)['durasi_menit'].mean().unstack(fill_value=0)
````

````python
# DEMO D4 — jumlah kunjungan per bulan
df['tanggal'] = pd.to_datetime(df['tanggal'])

kunjungan_bulan = (
    df.groupby(df['tanggal'].dt.month)
      .size()
      .sort_values(ascending=False)
)

print('Jumlah kunjungan per bulan:')
print(kunjungan_bulan)
````

````python
# TODO D5 — satu temuan deskriptif
# Ambil prodi dengan rata-rata kepuasan tertinggi dan terendah.
kepuasan_prodi = df.groupby('prodi')['kepuasan'].mean().sort_values()

prodi_terendah = kepuasan_prodi.index[0]
nilai_terendah = kepuasan_prodi.iloc[0]
prodi_tertinggi = kepuasan_prodi.index[-1]
nilai_tertinggi = kepuasan_prodi.iloc[-1]

print(
    f'{prodi_terendah} memiliki rata-rata kepuasan terendah '
    f'sebesar {nilai_terendah:.2f}, sedangkan {prodi_tertinggi} '
    f'memiliki rata-rata kepuasan tertinggi sebesar {nilai_tertinggi:.2f}.'
)
````

## Bagian F — Checkpoint proyek (dikerjakan bersama tim, di luar kelas)

Bagian ini menerapkan pemeriksaan Bagian A pada **dataset calon proyek tim**. Dataset tim yang tersedia di sini adalah **Data Stunting Indonesia**, terpisah dari dataset perpustakaan yang digunakan di Bagian A–E.

````python
# Muat dataset tim Data Stunting Indonesia
FILE_TIM = 'Data Stunting Indonesia.csv'

if not os.path.exists(FILE_TIM):
    try:
        from google.colab import files
        uploaded_tim = files.upload()
        if FILE_TIM not in uploaded_tim and len(uploaded_tim) > 0:
            FILE_TIM = next(
                (nama for nama in uploaded_tim if 'stunting' in nama.lower()),
                next(iter(uploaded_tim))
            )
    except ImportError:
        raise FileNotFoundError(
            'Letakkan file Data Stunting Indonesia CSV di folder notebook.'
        )

df_tim = pd.read_csv(FILE_TIM, sep=';', decimal=',', encoding='utf-8-sig')
df_tim = df_tim.loc[:, ~df_tim.columns.astype(str).str.match(r'^Unnamed')]
df_tim.columns = df_tim.columns.astype(str).str.strip()
df_tim['Provinsi'] = df_tim['Provinsi'].astype(str).str.strip()

kolom_tahun_tim = [kolom for kolom in df_tim.columns if str(kolom).isdigit()]
for kolom in kolom_tahun_tim:
    df_tim[kolom] = pd.to_numeric(df_tim[kolom], errors='coerce')

print('Baris, kolom:', df_tim.shape)
df_tim.info()
````

````python
# Pemeriksaan kualitas dataset tim
print('Lima baris awal:')
display(df_tim.head())

print('\nMissing value per kolom:')
print(df_tim.isnull().sum())

print('\nJumlah baris duplikat:', df_tim.duplicated().sum())
print('\nJumlah provinsi unik:', df_tim['Provinsi'].nunique())
print('\nRingkasan numerik:')
display(df_tim.describe())
````
